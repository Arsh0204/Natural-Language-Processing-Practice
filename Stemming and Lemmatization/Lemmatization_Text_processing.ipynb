Lemmatization is the process of converting a word into its correct base or dictionary form. In simple words: Lemmatization finds the meaningful, correct root word (dictionary form) of a word.

Lemmatization techniques is like stemming, The output we will get after lemmatizationis calles lemma which is a root word rather than root stem, after lemmatization we will be getting avalid word that means the same thing

```python
from nltk.stem import WordNetLemmatizer
'''
POS - Noun-n
verb-v
adjective-a
adverb-r
'''
#these are arguments with lemmatization
```

```python
lemmatizer = WordNetLemmatizer()
```

```python
print(lemmatizer.lemmatize('going'))
print(lemmatizer.lemmatize('history'))
```

```python
#pos -v (verb)
print(lemmatizer.lemmatize('going', pos='v'))
```

```python
#pos -r (adverb)
print(lemmatizer.lemmatize('going', pos='r'))
```

```python
#pos -a (adjective)
print(lemmatizer.lemmatize('going', pos='a'))
```

#out of these verb is correct, because going ka go hona cahiye root word 

```python
words = ["sleeping","slept","washing","cried","baked","broke","ate","crying","finally","history","writes","program"]
```

```python
#by default pos is n, which is noun
#Noun: A noun is the name of a person, place, thing, or idea.
#below these words are verbs, so we need to use pos = v
for word in words:
    print(word+"--->"+lemmatizer.lemmatize(word,))
```

```python
#A verb is a word that shows an action or a state of being.
#for verb, we have passed pos='v' that means we want to convert it to verb
for word in words:
    print(word+"--->"+lemmatizer.lemmatize(word, pos='v'))
```

all these words are verbs
