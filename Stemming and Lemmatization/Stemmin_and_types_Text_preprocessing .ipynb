```python
#for classification problem we use stemming, like classify email spam or ham
#Types of stemming techniques
#1 PorterStemmer
from nltk.stem import PorterStemmer
```

```python
words = ["sleeping","slept","washing","cried","baked","broke","ate","crying","finally","history","writes","program"]
```

```python
stemming = PorterStemmer()
for word in words:
    print(word+"--->"+stemming.stem(word))
```

```python
#major disadvantage od stemming
print(stemming.stem('congratulation'))
print(stemming.stem('study'))
print(stemming.stem('cry'))
print(stemming.stem('history'))
```

stemming works for good number of words, but not for rest of some words like example above, But we can fix this with the help of lemmitization

```python
#second method
#RegexpStemmer class
from nltk.stem import RegexpStemmer
```

```python
reg_stemmer = RegexpStemmer('ing$|s$|e$|able$', min=4)
#this means('ing$|s$|e$|able$') agar kisi word k piche ing,s,e,able use remove krdo, lets take some expales
```

```python
print(reg_stemmer.stem('eating'))
print(reg_stemmer.stem('comfortable'))
print(reg_stemmer.stem('sweets'))
print(reg_stemmer.stem('mangoes'))
```

```python
print(reg_stemmer.stem('ingeating'))
```

```python
# here $ means remove ing from last, if we remove dollar then ing can be remove from anywhere from word, chahe aage s
#here we remove dollar sign in argument from ing
reg_stemmer = RegexpStemmer('ing|s$|e$|able$', min=4)
print(reg_stemmer.stem('ingeating'))
```

now see ing removed from both suffix and prefix from a word, chahe khi bhi ho remove ho jayega

```python
#Another Method
#snowball Stemma, it gives better accuracy as compare to the porter stemma, accuracy means better form a word
from nltk.stem import SnowballStemmer
```

```python
snowball_stemmer = SnowballStemmer('english')
```

```python
for word in words:
    print(word+"--->"+snowball_stemmer.stem(word))
```

```python
#some comparison between porter stemming vs snowball stemming to check snowball stemming is better 
print("Porter stemming fairly --->",stemming.stem('fairly'))
print("snowball stemming fairly --->",snowball_stemmer.stem('fairly'))
```

```python
print("Porter stemming sportingly --->",stemming.stem('sportingly'))
print("snowball stemming sportingly --->",snowball_stemmer.stem('sportingly'))
```

in order to overcome the porter's stemming disadvantage we use snowball stemming
