```python
corpus = """Hello My name is Arsh Shamrma.
I am learning DL,NLP and Machine Learning.
I am also a sports player. I like to play! cricket."""
```

```python
print(corpus)
```

```python
#Tokenization 
#convert Paragraph ---> sentence
from nltk.tokenize import sent_tokenize
sent_tokenize(corpus)
```

It will return in tokens from paragraph to sentences or documents, it will return into list

```python
documents = sent_tokenize(corpus)
```

```python
documents
print("Type Of document",type(documents))
```

```python
#looping in documents, sentences
for sentence in documents:
    print(sentence)
```

there are 5 sentences

```python
# Tokenization
# convert Paragraph ---> words
# convert Sentence ---> words
from nltk.tokenize import word_tokenize
```

```python
#it will convert paragraph to words
word_tokenize(corpus)
```

here full stop, ! is treated as seperate word

```python
#converting sentence to word, i think there is no direct way
for sentence in documents:
    print(word_tokenize(sentence))
```

```python
from nltk.tokenize import TreebankWordTokenizer
```

```python
#another method to tokenize paragraph ---> word
tokenizer = TreebankWordTokenizer()
tokenizer.tokenize(corpus)
```

with the help of TreebankWordTokenizer() fullstop "." will not treated as seperate word but in word tokenize it will treated as seperate word, but last full stop will considered as seperate word
