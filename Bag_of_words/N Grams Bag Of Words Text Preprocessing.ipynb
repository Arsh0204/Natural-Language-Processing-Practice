```python
import pandas as pd 
sentences = pd.read_csv("SMSSpamCollection_2columns.csv")
```

```python
from nltk.stem import SnowballStemmer
snowball_stem = SnowballStemmer('english')
```

```python
from nltk.corpus import stopwords
import nltk
nltk.download('stopwords')
```

```python
import re
```

```python
corpus = []
for i in range(0,len(sentences)):
    review = re.sub('[^a-zA-z]', ' ', sentences['message'][i])
    review = review.lower()
    review = review.split()

    review = [snowball_stem.stem(word) for word in review if word not in stopwords.words('english')]
    review = ' '.join(review)
    corpus.append(review)
    
```

```python
#importing bag of words
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features = 200)
```

```python
X = cv.fit_transform(corpus).toarray()
```

```python
X
```

```python
cv.vocabulary_.
```

```python
# bag of words with n-gram, unigram(1,1)
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features = 200, ngram_range = (1,1))
X = cv.fit_transform(corpus).toarray()
```

```python
cv.vocabulary_
#nothing will change because by default cv model ngram_range=(1,1) which is unigram use kr rha hota h
```

```python
#bag of words,  combination of unigram and bigram ngram_range=(1,2)
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features = 500, ngram_range = (1,2))
X = cv.fit_transform(corpus).toarray()
```

```python
cv.vocabulary_
```

Now you will see there are 2-words(bigrams), but kam dikh rhe h because we gave max_features = 500, so top 500 max frequency wale words hi ayenge, shyd bi-gram words kam honge top 500 m

```python
#only bigrams(2,2), ignoring unu igrams 
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features = 500, ngram_range = (2,2))
X = cv.fit_transform(corpus).toarray()
```

```python
cv.vocabulary_
```

Now you will see only unigrams, because we gave the ngram_range(2,2) which means only unigram

```python
#only Trigrams(3,3), 
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features = 500, ngram_range = (3,3))
X = cv.fit_transform(corpus).toarray()
```

```python
cv.vocabulary_
```

now you will se trigrams, combinations of three words
