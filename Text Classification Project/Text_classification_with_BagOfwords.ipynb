```python
#now basic text preprocessing
#--Remove tags
#--lowercase
#--remove stopwords
import re 
import nltk
nltk.download('stopwords')
```

```python
def remove_tags(raw_text):
    cleaned_text = re.sub(re.compile('<.*?>'), '', raw_text)
    return cleaned_text
```

```python
df['review'] = df['review'].apply(remove_tags)
```

```python
df
```

```python
df['review'] = df['review'].apply(lambda x:x.lower())
```

```python
from nltk.corpus import stopwords
sw_list = stopwords.words('english')
df['review']= df['review'].apply(lambda x: [item for item in x.split() if item not in sw_list]).apply(lambda x:" ".join(x))
```

```python
#X and y ko divide krdiya, x is input(review) and y is positive or negative
X = df.iloc[:,0:1]
y = df['sentiment']
```

```python
#here (sentiment) positive or negative m h, hme 0 and 1 m chahiye because ml 0/1 ya true/false smajhta h
from sklearn.preprocessing import LabelEncoder
encoder = LabelEncoder()
y = encoder.fit_transform(y)
```

```python
#Train test split
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size = 0.2, random_state = 1)
```

```python
#Applying bag of words
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer()
```

```python
X_train_bow = cv.fit_transform(X_train['review']).toarray()
X_test_bow = cv.transform(X_test['review']).toarray()          #hmne yaha transform kiya only fit nhi kiya, means model ko nhi diya data
```

```python
X_train_bow.shape
# 48282 itne features bne h, hmaari vocabulary m 48282 words bne h
```

```python
X_test_bow.shape
# vocabulary m 48282 words bne h
```

```python
#now applying ml algorithm
from sklearn.naive_bayes import GaussianNB
gnb = GaussianNB()
```

```python
gnb.fit(X_train_bow,y_train)
```

```python
#hmne 20% jo x test data tha(review column wala) usko predict kiya
y_pred = gnb.predict(X_test_bow)
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix
accuracy_score(y_test, y_pred)   #y_test vo data h(sentiment column ka) jo x_test ka real output h and y_pred ml model n predict kiya h
```

```python
confusion_matrix(y_test, y_pred)
```

```python
reviews = [
    "I absolutely loved this movie, it was fantastic!",
    "The product quality is excellent and exceeded my expectations.",
    "The service was amazing, and the staff were very helpful.",
    "This is one of the best experiences I have ever had.",
    "The food was delicious and I really enjoyed every bite.",
    "I hated this movie, it was extremely boring.",
    "The product was terrible and broke after one day.",
    "The service was disappointing and the staff were rude.",
    "This was a complete waste of my time and money.",
    "The food tasted awful and I couldn't finish it."
]

# Convert reviews into TF-IDF
reviews_tfidf = tfidf.transform(reviews)

# Predict sentiment
predictions = rf.predict(reviews_tfidf)

# Print results
for review, prediction in zip(reviews, predictions):
    if prediction == 1:
        sentiment = "Positive 😊"
    else:
        sentiment = "Negative 😞"
        
    print(f"Review: {review}")
    print(f"Sentiment: {sentiment}")
    print("-" * 60)
```
