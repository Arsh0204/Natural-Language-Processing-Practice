```python
import gensim
import os
```

```python
#tokenize-corpus to sentence m convert krne k liye
from nltk.tokenize import sent_tokenize

#basic text_preprocessing k liye (simple_preprocess) jo hum pehle loop m krte the
from gensim.utils import simple_preprocess

story = []   #hme isme sbhi files ka data save krna h
for filename in os.listdir('Data'):           #loop laga diya data k andr ki files read krne ka
     if filename.endswith('.txt'):
        try:
            with open(os.path.join('Data', filename), encoding='utf-8') as f:
                corpus = f.read()
        except UnicodeDecodeError:
            with open(os.path.join('Data', filename), encoding='cp1252') as f:
                corpus = f.read()
                
     raw_sentence = sent_tokenize(corpus)             #corpus to sent tokenize
     for sent in raw_sentence:
        story.append(simple_preprocess(sent))                #story m sbhi sentence add krdiye aur text preprocess krke(removing stopwords,re etc)
```

```python
len(story)
```

```python
#2d list/array
#as we can see every sentence is tokenized
#story
```

```python
#model training starts
```

```python
model = gensim.models.Word2Vec(
    window=10,     #we take window size = 10 means 5words left 5 right bich wala output
    min_count=2,
    workers=4,
)
```

```python
model.build_vocab(story)
```

```python
model.train(story, total_examples=model.corpus_count, epochs = model.epochs)
```

Finally our model is trained

```python
#ab sbhi methods laga aur words k relations/ similarity percentage dekh
```
