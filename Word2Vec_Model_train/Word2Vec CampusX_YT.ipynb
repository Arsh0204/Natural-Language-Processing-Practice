```python
import gensim.downloader as api

wv = api.load('word2vec-google-news-300')
#model is stored in wv
```

```python
#word2vec advantage jisme vo similar word find out kr leta h kisi bhi word ka jo corpus m ho
#jaise happy ka similar joy etc, y dono words ka relation understand krta h dhundh bhi deta h

#y pehle man word ka vector find out krega then, us vector k aas paas(closest) waale vectors k words ko output m dega

wv.most_similar('man')
```

```python
wv.most_similar('man')
```

```python
# If we want to calculate the vector similarity ya distance between 2 words, 
# that unka relation or ya vo kitne similar h 

wv.similarity('man','woman')
```
